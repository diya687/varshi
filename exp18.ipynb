```python
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
print(df)
print(df.isnull().sum())
mean=df.mean()
print(mean)
df=df.fillna(mean)
print(df)
print('missing value after replacement:',df.isnull().sum())
print('first 5 rows',df.head())
```
